```python
#ARMSTRONG NUMBER
d=0
r=0
n = int(input("enter a number"))
x = n
x1 = n
digit = 0
while x1>0:
    x1=x1//10
    digit+=1
while n > 0:
    d = n%10
    r=r+(d**digit)
    n=n//10
if r==x:
    print("armstrong num")
else:
    print("not armstrong")
```
