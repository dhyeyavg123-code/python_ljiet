```python
sec = int(input("enter number of seconds: "))
sec_in_year = 3600*24*360
sec_in_month = 3600*24*30
sec_in_day = 3600 * 24
y = sec // sec_in_year
temp = sec % sec_in_year
m = temp // sec_in_month
temp1 = temp % sec_in_month
d = temp1 // sec_in_day
tempp = temp1 % sec_in_day
hour = tempp // 3600
remsec = tempp % 3600
minute = remsec // 60
fsec = remsec % 60
print(y , "years, ", m , "months, " , d ,"days, ",  hour , " hours, " , minute , "minutes and " , fsec , "seconds")
```
